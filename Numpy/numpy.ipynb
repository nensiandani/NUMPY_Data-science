```python
a = [100,200,300,400,500,600]
b = [10,20,30,40,50,60]

price = np.array(a)
quantity = np.array(b)

#print(np.sum(price))
print(price,"\n" ,quantity)
print()
c = np.cumprod([price,quantity], axis = 0)
print(c[1].sum())
```
